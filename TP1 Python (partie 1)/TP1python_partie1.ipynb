# 1. Rappel de Cours : Opérations de conversions

En Python, il est très fréquent de devoir transformer le type d'une donnée pour effectuer certains calculs ou affichages. C'est ce que l'on désigne par les opérations de conversions.

La fonction `input()` renvoie toujours une chaîne de caractères (`str`). Pour manipuler mathématiquement ces saisies, ou inversement pour concaténer des nombres avec du texte, il faut utiliser des fonctions intégrées de conversion explicite (aussi appelées *casting*) :

- `int(x)` : convertit la valeur `x` en un nombre entier.
- `float(x)` : convertit la valeur `x` en un nombre à virgule flottante (réel).
- `str(x)` : convertit la valeur `x` en une chaîne de caractères.

````python
note_saisie = "16.5"

note_reelle = float(note_saisie)

print(f"Valeur convertie : {note_reelle}, Type actuel : {type(note_reelle)}")

coefficient = 3
message_affichage = "Le coefficient de ce module d'informatique est " + str(coefficient)
print(message_affichage)
````

## Exercice de Mathématiques : Analyse d'une équation

Écrivez un script Python qui demande à l'utilisateur de saisir trois nombres réels $a$, $b$ et $c$ représentant les coefficients d'une équation de la forme $ax^2 + bx + c = 0$.

Votre programme doit respecter **strictement** l'architecture suivante :

1. **Première partie (Tests imbriqués) :** 
   Vérifiez si $a$ est égal à $0$. 
   - Si c'est le cas, faites un test imbriqué pour vérifier si $b$ est également égal à $0$. Affichez un message approprié (ex: "Équation dégénérée" ou "Équation du premier degré") et changez une variable booléenne `est_second_degre` à `False`.
   - Sinon, définissez `est_second_degre` à `True`.

2. **Deuxième partie (Autre structure conditionnelle) :**
   À la suite de ce premier bloc (et non à l'intérieur), créez une nouvelle structure conditionnelle qui s'exécute uniquement si `est_second_degre` est vrai. 
   - Calculez le discriminant $\Delta = b^2 - 4ac$.
   - Déterminer et afficher le nombre de solutions réelles de l'équation selon le signe de $\Delta$.

````python
a = float(input("Coefficient a : "))
b = float(input("Coefficient b : "))
c = float(input("Coefficient c : "))

if a == 0:
    if b == 0:
        print("Équation dégénérée (a = 0 et b = 0).")
    else:
        print("Équation du premier degré (a = 0).")
    est_second_degre = False
else:
    est_second_degre = True

if est_second_degre:
    delta = b**2 - 4 * a * c
    print(f"Discriminant : Δ = {delta}")
    if delta > 0:
        print("Δ > 0 : l'équation admet 2 solutions réelles distinctes.")
    elif delta == 0:
        print("Δ = 0 : l'équation admet 1 solution réelle (double).")
    else:
        print("Δ < 0 : l'équation n'admet aucune solution réelle.")
````

# 3. Les boucles en Python : for et while

Les boucles permettent de répéter l'exécution d'un bloc d'instructions. En Python, le début du bloc à répéter est toujours signalé par le symbole deux-points (`:`) à la fin de la ligne, et le contenu du bloc doit obligatoirement être **indenté**.

## 1. La boucle `for` et la fonction `range()`

La boucle `for` est utilisée lorsqu'on connaît à l'avance le nombre de répétitions ou pour parcourir une séquence. Elle utilise le mot-clé `in` pour affecter tour à tour les valeurs de la séquence à une variable. La fonction `range()` est très souvent utilisée avec `for` pour générer une suite de nombres entiers.

**Syntaxe :**
```python
for variable in range(valeur_debut, valeur_fin, pas):
    # bloc d'instructions à répéter

### Les règles importantes de la fonction `range()` en Python

La fonction `range()` permet de générer une séquence de nombres entiers. Sa syntaxe complète est `range(valeur_debut, valeur_fin, pas)`. 

Voici les trois règles fondamentales à retenir concernant ses paramètres :

*   **La valeur de début (incluse) :** C'est le point de départ de la séquence. Si elle n'est pas spécifiée explicitement (comme dans `range(5)`), sa valeur par défaut est toujours **`0`**. 
*   **La valeur de fin (strictement exclue) :** C'est la limite supérieure (ou inférieure) de la séquence. **Cette valeur n'est jamais prise en compte** dans les nombres générés. La séquence s'arrête toujours juste avant cette limite (ex: `range(0, 5)` générera 0, 1, 2, 3, 4).
*   **Le pas (l'incrément) :** Il représente l'écart ou le saut entre chaque nombre de la séquence. S'il est omis (comme dans `range(1, 10)`), sa valeur par défaut est **`1`**. Il est tout à fait possible d'utiliser un pas négatif pour créer un compte à rebours (ex: `range(10, 0, -1)`).

**Exemples récapitulatifs :**
- `range(3)` génère : `0, 1, 2` *(début=0, fin=3, pas=1)*
- `range(2, 6)` génère : `2, 3, 4, 5` *(début=2, fin=6, pas=1)*
- `range(0, 10, 2)` génère : `0, 2, 4, 6, 8` *(début=0, fin=10, pas=2)*

## 2. La boucle `while`

La boucle `while` (qui se traduit par "tant que") est utilisée pour répéter un bloc d'instructions de manière indéfinie, `tant qu'une` condition logique est évaluée à `True` (Vraie).

```python
# Syntaxe
while condition:
    # bloc d'instructions à répéter
    # Attention : il faut obligatoirement inclure ici une opération 
    # qui modifiera la condition pour qu'elle devienne False, 
    # sinon on crée une boucle infinie.
```

## Exercice : Contrôle de saisie et boucle d'affichage

Écrivez un script Python qui demande à l'utilisateur de saisir un **nombre entier impair**. 

Votre programme doit accomplir les tâches suivantes :

1. **Validation de la saisie (Boucle `while`) :** 
   Le programme doit vérifier la valeur saisie. Tant que l'utilisateur entre un nombre pair ou un nombre réel (décimal), affichez un message d'erreur explicite et redemandez-lui une nouvelle saisie. La boucle ne doit s'arrêter que lorsqu'un entier impair valide est fourni.

2. **Génération de la séquence (Boucle `for`) :** 
   Une fois la condition remplie, utilisez une boucle `for` couplée à la fonction `range()` pour afficher tous les nombres entiers en partant de $0$ jusqu'au nombre saisi (**inclus**).

**Exemple de déroulement attendu :**
> *Veuillez saisir un nombre entier impair : 4*
> *Erreur : 4 est un nombre pair. Veuillez réessayer : 3.5*
> *Erreur : Vous avez saisi un nombre décimal. Veuillez réessayer : 5*
> *Saisie acceptée ! Voici les nombres de 0 à 5 :*
> *0*
> *1*
> *2*
> *3*
> *4*
> *5*

````python
saisie = float(input("Veuillez saisir un nombre entier impair : "))

while saisie != int(saisie) or saisie % 2 == 0:
    if saisie != int(saisie):
        saisie = float(input("Erreur : Vous avez saisi un nombre décimal. Veuillez réessayer : "))
    else:
        saisie = float(input(f"Erreur : {int(saisie)} est un nombre pair. Veuillez réessayer : "))

nombre = int(saisie)

print(f"Saisie acceptée ! Voici les nombres de 0 à {nombre} :")
for i in range(nombre + 1):
    print(i)
````

# 4. Rappel de Cours : Les Fonctions en Python

Une fonction est un bloc d'instructions réutilisable, conçu pour effectuer une tâche précise.

## 1. L'immuabilité des types de base
En Python, les types de base (`int`, `float`, `bool`, `complex`, `str`) sont **immuables** (non modifiables sur place). 
Lorsqu'une variable de ce type est passée en paramètre à une fonction, toute tentative de modification à l'intérieur de la fonction va en réalité créer un **nouvel objet local**. La variable d'origine située à l'extérieur de la fonction restera totalement intacte.

## 2. Les paramètres par défaut
Lors de la définition d'une fonction, il est possible d'attribuer une valeur par défaut à un ou plusieurs paramètres à l'aide de l'opérateur `=`. Si l'utilisateur ne fournit pas cet argument lors de l'appel de la fonction, c'est la valeur par défaut qui sera utilisée.
```python
def saluer(nom, message="Bonjour"):
    print(f"{message} {nom}")

## Exercice : Traitement des notes

1. **Paramètre par défaut & Immuabilité :** Créez une fonction `ajouter_bonus(note, bonus=2)` qui retourne la note augmentée du bonus. Déclarez une variable `x = 10`, appelez la fonction avec `x`, puis affichez `x` pour prouver que sa valeur initiale est restée intacte.
2. **`map()` :** À partir de la liste `notes = [12.0, 15.5]`, utilisez la fonction `map()` pour créer une nouvelle liste contenant ces notes augmentées du bonus par défaut.
3. **`zip()` :** Avec la liste `etudiants = ["Ali", "Sara"]`, utilisez une boucle `for` et `zip()` pour afficher chaque étudiant avec sa nouvelle note.

````python
def ajouter_bonus(note, bonus=2):
    note = note + bonus
    return note

x = 10
resultat = ajouter_bonus(x)
print(f"Valeur retournée : {resultat}")
print(f"x après l'appel : {x}  -> la valeur initiale est restée intacte")

notes = [12.0, 15.5]
nouvelles_notes = list(map(ajouter_bonus, notes))
print(f"Nouvelles notes : {nouvelles_notes}")

etudiants = ["Ali", "Sara"]
for etudiant, note in zip(etudiants, nouvelles_notes):
    print(f"{etudiant} : {note}")
````

# 5. Rappel de Cours : Les Listes en Python

Une liste en Python est une structure de données qui permet de stocker une collection ordonnée d'éléments. Contrairement aux types de base vus précédemment, les listes sont **mutables**, ce qui signifie que leur contenu peut être modifié après leur création.

**1. Création et accès aux éléments**
Les listes sont définies par des crochets `[]`, et les éléments sont séparés par des virgules.
L'accès se fait par un indice, en sachant que le premier élément est à l'indice **0**.
```python
notes = [14.5, 12.0, 18.5, 9.0]
premiere_note = notes[0]   # 14.5
derniere_note = notes[-1]  # 9.0 (l'indice -1 part de la fin)

## Exercice : Saisie ordonnée et calcul de la médiane

1. Demandez à l'utilisateur de saisir le **nombre total de notes** qu'il souhaite traiter.
2. Utilisez une boucle `for` pour lui demander de saisir chaque note une par une, avec une contrainte stricte : les notes doivent être saisies **par ordre croissant**. Stockez ces notes dans une liste. *(Indice : utilisez une boucle `while` pour forcer l'utilisateur à resaisir si la nouvelle note est inférieure à la précédente).*
3. Déterminez si le nombre de notes est pair ou impair.
4. **Calculez et affichez la médiane** à partir de cette liste déjà triée lors de la saisie.

````python
n = int(input("Nombre total de notes : "))

notes = []
for i in range(n):
    note = float(input(f"Note n°{i + 1} : "))
    while i > 0 and note < notes[-1]:
        note = float(input(f"Erreur : la note doit être >= {notes[-1]}. Note n°{i + 1} : "))
    notes.append(note)

print(f"Notes saisies : {notes}")

milieu = n // 2
if n % 2 == 0:
    print(f"Le nombre de notes ({n}) est pair.")
    mediane = (notes[milieu - 1] + notes[milieu]) / 2
else:
    print(f"Le nombre de notes ({n}) est impair.")
    mediane = notes[milieu]

print(f"La médiane est : {mediane}")
````

# 5. Cours : Structuration de données pour l'Intelligence Artificielle Agentique en Python

En programmation de systèmes multi-agents, Python propose trois structures de données natives principales pour regrouper, étiqueter et organiser la mémoire et le contexte des agents.

*   **Les Listes `[ ]` :** Des collections ordonnées et *mutables* (modifiables après création). Elles sont idéales pour stocker une séquence chronologique qui évolue dynamiquement, comme l'historique d'une conversation ou la suite des étapes de raisonnement (chaîne de pensées) d'un agent.
    *   *Exemple :* `historique_actions = ["observation", "reflexion", "appel_outil", "reponse_finale"]`
*   **Les Tuples `( )` :** Des collections ordonnées mais *immuables* (non modifiables). Ils sont parfaits pour regrouper des valeurs intrinsèquement liées qui ne doivent jamais être altérées par erreur lors du cycle d'exécution, comme l'association stricte entre le nom d'un outil et son identifiant, ou les coordonnées de l'agent dans un environnement.
    *   *Exemple :* `signature_outil = ("WebSearch_API", "v2.1", "lecture_seule")` *(nom, version, permission)*
*   **Les Dictionnaires `{ }` :** Des collections qui stockent les données sous forme de paires `clé: valeur`. Ils permettent d'accéder aux informations via des étiquettes textuelles explicites. C'est la structure indispensable pour définir le profil d'un agent, ses paramètres de modèle (LLM), ou son état cognitif global.
    *   *Exemple :* `configuration_agent = {"role": "analyste", "modele": "gpt-4", "temperature": 0.2, "memoire_active": True}`

## Problème Pratique : Configuration d'un système multi-agents (IA)
